```python
# Import libraries
import pandas as pd

# Make a DataFrame
df = pd.DataFrame({'A': [0, 1, 2, 9, 4],
                   'B': [4, 9, 2, 0, 3],
                   'C': [8, 2, 7, 6, 8]})

# Perform the rolling window calculations with window size = 2
roll_mean = df.rolling(window = 2).mean()
print(roll_mean)
```
