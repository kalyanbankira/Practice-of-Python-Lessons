```python
# 1. Import the 'numpy', 'pandas' and 'time' modules
import numpy as np
import pandas as pd
import time

# 2. Record the time before creating a Pandas Series.
py_t0 = time.time()

# 3. Create a Python list containing 100,000 items from 1 to 100,000.
py_series = pd.Series([i for i in range(1, 100001)])

# 4. Record the time after creating the list.
py_t1 = time.time()

# 5. Calculate the time taken to create a Python list by computing the time difference.
py_time_diff = (py_t1 - py_t0)
print("Time taken to create a Pandas series is", py_time_diff, "seconds.")

# 6. Record the time before creating a NumPy array.
np_t0 = time.time()

# 7. Create a NumPy array containing just one row and 100,000 items from 1 to 100,000.
np_ones = np.arange(1, 100001)

# 8. Record the time after creating the NumPy array.
np_t1 = time.time()

# 9. Calculate the time taken to create a NumPy array by computing the time difference.
np_time_diff = (np_t1 - np_t0)
print("Time taken to create a NumPy array is", np_time_diff, "seconds.")

# 10. Calculate the factor by which a NumPy array creation is faster than a Python list creation.
print("A NumPy array is", py_time_diff // np_time_diff, "times faster than a Pandas Series for the same size.")
# Recall that // means floor division.
```

```python
# Solution
import pandas as pd
import random
def compute_median(num_series):
  result = 0
  num_list = [num for num in num_series]
  num_list.sort()
  if len(num_list) % 2 == 0:
    result = (num_list[(len(num_list) // 2) - 1] + num_list[len(num_list) // 2]) / 2
  else:
    result = float(num_list[len(num_list) // 2])
  return result

num_series = pd.Series([random.randint(20, 40) for i in range(9)])
print("Median without using the standard median() function:", compute_median(num_series))
print("Median using the standard median() function:", num_series.median())
```

```python
# Solution
def time_saved(s_lim, s_avg, d):
    norm_time = (d / s_lim) * 60
    fast_time = (d / s_avg) * 60
    worth = norm_time - fast_time
    return worth
time_saved(80, 90, 4000)
```

```python
# Create a function to normalise a Pandas series using the mean normalisation method.
import pandas as pd
def mean_normalise(series):
  norm_series = (series - series.mean()) / (series.max() - series.min())
  return norm_series

test_series = pd.Series([7, 11, 2000, 119, 540])
mean_normalise(test_series)
```
