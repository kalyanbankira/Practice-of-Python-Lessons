```python
# Print n -digit number
def num_generator(n):
  req_num=0
  count=n - 1
  while count>=1:
    req_num=req_num + (10** count) * n
    count=count -1
  return req_num
for n in range(1,10):
    print(num_generator(n), "=", n)
```
